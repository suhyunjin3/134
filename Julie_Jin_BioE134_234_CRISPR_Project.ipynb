CRISPR-Cas9 is a gene-editing technology that enables precise modifications to DNA at specific locations in the genome. This technology uses two key components: the Cas9 enzyme and a guide RNA (gRNA). Here's a breakdown of their roles:

**Cas9**: A protein that acts as a pair of molecular scissors, capable of cutting DNA strands. Its activity is directed by the gRNA, allowing it to target specific DNA sequences.

**Guide RNA (gRNA)**: A short, synthetic RNA sequence designed to bind to a specific DNA sequence in the genome. The gRNA guides the Cas9 to the exact location where a cut is needed. This targeting is achieved through complementary base pairing between the gRNA and the target DNA sequence.
The Cas9-gRNA complex locates and binds to a specific site in the DNA called the Protospacer Adjacent Motif (PAM) with the sequence 'NGG'.  The PAM site is essential for Cas9 recognition and binding. Once bound, Cas9 makes a double-strand break in the DNA, enabling gene editing through mechanisms such as knock-out, insertion, or replacement.


## 1) Writing and invoking a validation function

Our program will be written to accept a string as the input DNA sequence. However, only strings made of A, T, C, and G are meaningful inputs. A general principle with the algorithms in this course is that they should only return results for valid inputs.  So, if an invalid string such as 'my_sweater_not_ATCG' were passed into the function, it should respond with an error indicating the input is invalid, terminating the thread.

Below we define a function named `validate_DNA` that takes a single string argument and returns true if it is a valid DNA sequence, and false if it does not.  In colab, functions and variables declared in earlier blocks remain accessible to later code.  So, you can use validate_DNA in future blocks of this notebook.

````python
def validate_DNA(seq):
    """
    Validates whether a given string contains only valid DNA nucleotides (A, T, C, G).

    Args:
        seq (str): The input string to be checked. This string can contain any characters,
                   and the function will determine if it exclusively contains 'A', 'T', 'C', and 'G'.

    Returns:
        bool: Returns True if the string contains only 'A', 'T', 'C', and 'G';
              otherwise, returns False.
    """
    # Return False if seq is None (meaning there is no string, it is null, which is different from saying an empty string like "")
    if seq is None:
        return False

    # Define a set of valid nucleotides for DNA
    valid_nucleotides = {'A', 'T', 'C', 'G'}

    # Check if every character in the string is a valid nucleotide
    return all(nucleotide in valid_nucleotides for nucleotide in seq)
````

After writing a function, the next thing to do is run it on a simple valid input, such as:

````python
# Test the function on a simple valid input
result = validate_DNA("ATCAGAGATCAG")
print(result)
````

## 2) Find the sequence "GG"
CRISPR-Cas9 targets a DNA sequence that includes a 20 bp protospacer followed immediately by the PAM sequence "NGG." To find a valid protospacer, the initial task is to locate a valid PAM sequence within the target DNA. The 'N' in "NGG" can be any of the four DNA bases, so practically, you only need to search for the "GG" sequence. String operations can be used to find the index of this PAM sequence.

I've already linked a call to validate_DNA as a guard clause. A guard clause is a preliminary check that ensures the input meets certain conditions before the rest of the function is executed. In this case, the guard verifies that the DNA sequence is valid (i.e., contains only 'A', 'T', 'C', and 'G'). If the input is invalid, the program will immediately throw an error, preventing further execution.

Another complication here is that the test sequence provided has a very early 'GG' which is too close to the end of the sequence to have a protospacer upstream of it.  Thus, your algorithm will need to ignore the first 23 or so bases of the sequence.

````python
def find_PAM(target):
    """
    Finds the position of the first occurrence of the 'GG' PAM sequence in a DNA target sequence.

    Args:
        target (str): The DNA sequence to be searched. The input string must consist of valid DNA
                      nucleotides ('A', 'T', 'C', 'G').

    Returns:
        int: The index of the first occurrence of 'GG' in the target sequence after the 23rd position.

    Raises:
        ValueError: If the target sequence is invalid or if no 'GG' PAM sequence is found.
    """
    if not validate_DNA(target):
        raise ValueError("Invalid target sequence")

    # TODO:  Complete this function
    pam_index = target.find("GG", 23)

    if pam_index == -1:
        raise ValueError("No GG PAM sequence found")

    return pam_index

# Test it out
print(find_PAM('CCCTAGATGCCTGGCTCAGAAACCTGCCAGTTTGCTGGCACGTTTTTTTCTTTTGTCTTTAGTTCTCACGTTTGTCATACTTGACAACGCTTCTTTAACCAAATATAATTGTTC'))
````

## 3) Find the Protospacer
After finding the PAM sequence, the protospacer will be the 20 bp upstream of it on the target. For the example sequence above, the first viable sequence is 'TCAGAAACCTGCCAGTTTGC', so that is what your code below should print out. To do this, you will need to use your string operations to slice the sequence from some start and end indices calculated relative to the pam_index.

````python
def find_protospacer(target):
    """
    Identifies the protospacer sequence within a DNA target sequence based on the location of the PAM sequence.

    The protospacer is a 20 base pair sequence that immediately precedes the PAM sequence ('NGG'). This function
    locates the PAM sequence using the `find_PAM` function and then extracts the 20 bp protospacer sequence
    located just before it.

    Args:
        target (str): The DNA sequence to be searched. The input string must consist of valid DNA
                      nucleotides ('A', 'T', 'C', 'G').

    Returns:
        str: The 20 base pair protospacer sequence found immediately before the PAM sequence.

    Raises:
        ValueError: If the target sequence is invalid or if no valid PAM sequence is found.
    """
    # Find the index of the PAM sequence using the find_PAM function
    pam_index = find_PAM(target)

    # TODO:  Complete this function
    protospacer = target[pam_index-21:pam_index-1]

    return protospacer

# Test it out
print(find_protospacer('CCCTAGATGCCTGGCTCAGAAACCTGCCAGTTTGCTGGCACGTTTTTTTCTTTTGTCTTTAGTTCTCACGTTTGTCATACTTGACAACGCTTCTTTAACCAAATATAATTGTTC'))
````

## 4) Compute the full gRNA sequence
If our goal is to chemically synthesize the guide RNA molecule, then you will need a complete sequence of it.  A Cas9 gRNA has the protospacer on its 5' end, and then this trcrRNA sequence on its 3' end:

```
gttttagagctagaaatagcaagttaaaataaggctagtccgttatcaacttgaaaaagtggcaccgagtcggtgc
```
You will need to combine those sequences together to get the full length sequence.

````python
def design_cas9_DNA(target):
    """
    Designs a DNA sequence for use in CRISPR-Cas9 gene editing by combining a protospacer sequence
    with a standard gRNA scaffold.

    This function first identifies the 20 base pair protospacer sequence immediately preceding the PAM
    sequence ('GG') within the provided DNA target sequence. It then appends a predefined gRNA scaffold
    sequence to the 3' end of the protospacer to create the full Cas9 guide RNA sequence.

    Args:
        target (str): The DNA sequence to be searched. The input string must consist of valid DNA
                      nucleotides ('A', 'T', 'C', 'G').

    Returns:
        str: A DNA sequence that includes the protospacer followed by the gRNA scaffold sequence.

    Raises:
        ValueError: If the target sequence is invalid or if no valid PAM sequence is found.
    """
    # Find the protospacer sequence within the target DNA
    protospacer = find_protospacer(target)

    # Define the gRNA scaffold sequence
    trcrRNA = 'gttttagagctagaaatagcaagttaaaataaggctagtccgttatcaacttgaaaaagtggcaccgagtcggtgc'.upper()

    # TODO:  Complete this function
    guide = ''.join([protospacer, trcrRNA])

    # Return the full Cas9 guide sequence
    return guide

# Test it out
print(design_cas9_DNA('CCCTAGATGCCTGGCTCAGAAACCTGCCAGTTTGCTGGCACGTTTTTTTCTTTTGTCTTTAGTTCTCACGTTTGTCATACTTGACAACGCTTCTTTAACCAAATATAATTGTTC'))
````
